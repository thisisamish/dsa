```python
import json
```

```python
creatorSheetsMap = {
    "striver": ["a2z", "sde", "75", "79"],
    "lb": ["450"],
    "neetcode": ["75", "150", "250"],
    "leetcode": ["75", "150"]
}
```

```python
merged = []
seen = set()
total_items = 0
items_count_map = {}
for creator, sheets in creatorSheetsMap.items():
    for sheet in sheets:
        skipped_count = 0
        with open(f"../deduped-data/{creator}/{sheet}.json", 'r', encoding="utf-8") as file:
            raw = json.load(file)
        total_items += len(raw)
        for item in raw:
            id = item['id']
            if id not in seen:
                merged.append(item)
                seen.add(id)
            else:
                skipped_count += 1
            items_count_map[id] = items_count_map.setdefault(id, 0) + 1
        print(f"Skipped {skipped_count} items from {creator}'s {sheet} sheet.")
```

```python
print(len(merged))
print(total_items)
print(f"Dupe percentage: {(((total_items - len(merged)) * 100) / total_items):.2f}%")
```

```python
# Find all those items which appear in all the 10 sheets
appear_in_all = []
for key, value in items_count_map.items():
    if value >= 8:
        appear_in_all.append(key)
```
