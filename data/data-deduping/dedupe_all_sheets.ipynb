```python
import json
```

```python
creatorSheetsMap = {
    "striver": ["a2z", "sde", "75", "79"],
    "lb": ["450"],
    "neetcode": ["75", "150", "250"],
    "leetcode": ["75", "150"]
}
```

```python
def findDupes(creator: str, sheetname: str) -> dict[str, int | str, list[list[dict[str, str]]]]:
    print(f"Extracting dupes data from {creator}'s {sheetname} sheet...")
    try:
        with open(f"../cleaned-data/{creator}/{sheetname}.json", encoding='utf-8') as file:
            raw = json.load(file)
    except Exception as e:
        raise Exception(f"Error while loading cleaned data of {creator}'s {sheetname} sheet: {e}")
    id_item_map = {}
    for item in raw:
        id_item_map.setdefault(item['id'], []).append(item)
    dupes = []
    total_dupes = 0
    total_items_with_dupes = 0
    for val in id_item_map.values():
        if len(val) > 1:
            total_dupes += len(val)
            total_items_with_dupes += 1
            dupes.append(val)
    dupes_data = {}
    dupes_data["dupes_data"] = dupes
    dupes_data["total_items_with_dupes"] = total_items_with_dupes
    dupes_data["total_dupes"] = total_dupes
    dupes_data["total_items_in_sheet_before_dedupe"] = len(raw)
    dupes_data["total_items_in_sheet_after_dedupe"] = len(raw) - total_dupes + total_items_with_dupes
    return dupes_data
```

```python
dupes_data_all_sheets = {}
for creator, sheets in creatorSheetsMap.items():
    for sheet in sheets:
        creator_data = dupes_data_all_sheets.setdefault(creator, {})
        creator_data[sheet] = findDupes(creator, sheet)
```

```python
with open("dupes_data_all_sheets.json", 'w', encoding='utf-8') as f:
    json.dump(dupes_data_all_sheets, f, ensure_ascii=False, indent=2)
```

```python
def validateDedupe(creator: str, sheetname: str, dupes_data_all_sheets: dict[str, int | str, list[list[dict[str, str]]]]) -> bool:
    print(f"Validating dedupe of {creator}'s {sheetname} sheet...")
    try:
        with open(f"../deduped-data/{creator}/{sheetname}.json", encoding='utf-8') as file:
            raw = json.load(file)
    except Exception as e:
        raise Exception(f"Error while loading deduped data of {creator}'s {sheetname} sheet: {e}")
    return dupes_data_all_sheets[creator][sheetname]["total_items_in_sheet_after_dedupe"] == len(raw)
```

```python
for creator, sheets in creatorSheetsMap.items():
    for sheet in sheets:
        print(f"{creator}'s {sheet} sheet dedupe VALID.") if validateDedupe(creator, sheet, dupes_data_all_sheets) else print(f"{creator}'s {sheet} sheet dedupe INVALID.")
```
